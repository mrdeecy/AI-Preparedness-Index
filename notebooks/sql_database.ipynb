```python
import os
import pandas as pd
import re
from sqlalchemy import create_engine, URL
import psycopg
from dotenv import load_dotenv
load_dotenv()

DB_CONFIG = {
    "dbname":   os.getenv("DB_NAME",   "census_project"),
    "user":     os.getenv("DB_USER",     "project_gmt"),
    "password": os.getenv("DB_PASSWORD"),
    "host":     os.getenv("DB_HOST",     "localhost"),
    "port":     int(os.getenv("DB_PORT", "5432")),
}
print("Config loaded:", {k: v for k, v in DB_CONFIG.items() if k != 'password'})

url = URL.create(
    "postgresql+psycopg",
    username=DB_CONFIG["user"],
    password=DB_CONFIG["password"],
    host=DB_CONFIG["host"],
    port=DB_CONFIG["port"],
    database=DB_CONFIG["dbname"],
)
```

```python
# LOAD CLEANED DATA INTO POSTGRESQL

# PostgreSQL Connection Details
engine = create_engine(url, echo=False)

# Upload cleaned dataset to PostgreSQL
df = pd.read_csv('cleaned_us_census_data.csv')
df.to_sql(
    name='cleaned_us_census_data',
    con=engine,
    if_exists='replace',
    index=False
)

print("\nCleaned dataset uploaded to PostgreSQL successfully.")

# VERIFY DATABASE TABLE

query = """
SELECT *
FROM cleaned_us_census_data
LIMIT 5;
"""

verification = pd.read_sql(query, engine)

print("\nDatabase Verification:")
print(verification)

```

```python
# STEP 18: SAVE FINAL BACKUP FILE

df.to_excel(
    "cleaned_us_census_data_backup.xlsx",
    index=False
)

print("\nExcel backup saved successfully.")
```
